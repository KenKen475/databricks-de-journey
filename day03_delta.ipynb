```python
%sql
CREATE OR REPLACE TABLE workspace.bronze.orders_1998 AS
SELECT *
FROM samples.tpch.orders
WHERE o_orderdate >= '1998-01-01';

```

```python
%sql
SELECT COUNT(*) FROM workspace.bronze.orders_1998;   -- record this number
```

```python
%sql
DESCRIBE DETAIL workspace.bronze.orders_1998;    -- format, numFiles, sizeInBytes
```

```python
%sql
DESCRIBE HISTORY workspace.bronze.orders_1998;   -- one row per version
```

```python
%sql
-- Version 1: UPDATE
UPDATE workspace.bronze.orders_1998
SET o_orderstatus = 'F'
WHERE o_orderstatus = 'O' AND o_orderdate < '1998-02-01';
```

```python
%sql
-- Version 2: DELETE
DELETE FROM workspace.bronze.orders_1998
WHERE o_orderpriority = '5-LOW';
```

```python
%sql
-- Version 3: INSERT
INSERT INTO workspace.bronze.orders_1998
SELECT * FROM samples.tpch.orders
WHERE o_orderdate = '1997-12-31';
```

```python
%sql
DELETE FROM workspace.bronze.orders_1998;
```

```python
%sql
SELECT COUNT(*) FROM workspace.bronze.orders_1998;
```

```python
%sql
SELECT COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 3
```

```python
%sql
RESTORE TABLE workspace.bronze.orders_1998 TO VERSION AS OF 3;
```

```python
%sql
SELECT COUNT(*) FROM workspace.bronze.orders_1998;
```

```python
%sql
-- 1. What happened after the RESTORE?
DESCRIBE HISTORY workspace.bronze.orders_1998;

-- 2. Row count at every version (adjust to your latest version number)
SELECT 0 AS version, COUNT(*) AS row_count FROM workspace.bronze.orders_1998 VERSION AS OF 0
UNION ALL SELECT 1, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 1
UNION ALL SELECT 2, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 2
UNION ALL SELECT 3, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 3
UNION ALL SELECT 4, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 4
UNION ALL SELECT 5, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 5
UNION ALL SELECT 6, COUNT(*) FROM workspace.bronze.orders_1998 VERSION AS OF 6
ORDER BY version;
```
